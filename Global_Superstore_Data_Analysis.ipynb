# Data Analysis Using Pandas – Global Superstore

This project demonstrates five fundamental data-analysis tasks using Python and Pandas:

1. Load a dataset and display basic information.
2. Handle missing values and duplicate records.
3. Group data by category and calculate total revenue.
4. Sort data using multiple columns.
5. Create a correlation matrix for numerical columns.


## 1. Load the dataset and display basic information

```python
import pandas as pd

# Load the Orders sheet from the Global Superstore Excel file
df = pd.read_excel('global_superstore_2016(1).xlsx', sheet_name='Orders')

print('Dataset shape:', df.shape)
display(df.head())
df.info()
display(df.describe(include='all').T)
```

## 2. Handle missing values and duplicates

```python
# Check missing values and duplicate records
print('Missing values before cleaning:')
display(df.isnull().sum())

print('Duplicate rows before cleaning:', df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Postal Code is an identifier, so missing values are represented as 0
if 'Postal Code' in df.columns:
    df['Postal Code'] = df['Postal Code'].fillna(0)

print('Missing values after cleaning:')
display(df.isnull().sum())
print('Duplicate rows after cleaning:', df.duplicated().sum())
```

## 3. Group data by category and find total revenue

```python
# In this dataset, Sales is used as the revenue measure
category_revenue = (
    df.groupby('Category', as_index=False)['Sales']
      .sum()
      .rename(columns={'Sales': 'Total Revenue'})
      .sort_values('Total Revenue', ascending=False)
)

display(category_revenue)
```

## 4. Sort data by multiple columns

```python
# Sort Category alphabetically and Sales from highest to lowest
sorted_data = df.sort_values(
    by=['Category', 'Sales'],
    ascending=[True, False]
)

display(sorted_data[['Category', 'Product Name', 'Sales', 'Profit']].head(10))
```

## 5. Create a correlation matrix for numerical columns

```python
# Select numerical columns and calculate pairwise correlations
numeric_data = df.select_dtypes(include='number')
correlation_matrix = numeric_data.corr()

display(correlation_matrix.round(2))
```

## Conclusion

The dataset was loaded and inspected using Pandas. Duplicate records were removed, missing Postal Code values were handled, category-wise revenue was calculated, records were sorted using multiple columns, and a correlation matrix was generated for numerical variables.
