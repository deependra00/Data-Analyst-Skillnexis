# Clean a Messy Dataset in Pandas

This project demonstrates practical data cleaning and preprocessing using Pandas.

### Tasks
- Handle missing values
- Remove duplicate records
- Filter rows
- Create new columns
- Export the cleaned dataset


```python
import pandas as pd

# 1. Load the messy CSV
df = pd.read_csv('data.csv')
print('Original shape:', df.shape)
display(df.head())

# 2. Clean column names and whitespace
df.columns = (df.columns.astype(str).str.strip().str.lower().str.replace(r'[^\\w]+', '_', regex=True).str.strip('_'))
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype('string').str.strip().replace({'': pd.NA})

# 3. Remove duplicates
print('Duplicates before:', df.duplicated().sum())
df = df.drop_duplicates().copy()

# 4. Handle missing values
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].median())
    else:
        mode = df[col].mode(dropna=True)
        df[col] = df[col].fillna(mode.iloc[0] if not mode.empty else 'Unknown')

# 5. Create new columns
df['numeric_total'] = df['duration'] + df['pulse']
df['non_missing_count'] = df.notna().sum(axis=1)

# 6. Filter rows
df = df[df['non_missing_count'] > 0].copy()

# 7. Save cleaned dataset
df.to_csv('cleaned_dataset.csv', index=False)
print('Final shape:', df.shape)
display(df.head())
```

## Result

The final dataset is cleaned, filtered, enriched with derived columns, and exported as `cleaned_dataset.csv`.
